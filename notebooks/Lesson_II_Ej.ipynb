### 👾 ¡La invasión alienígena!
Escenario:

Eres analista de datos en la Organización de Defensa Terrestre (EDO). Acabas de recibir una transmisión con información sobre una flota alienígena que se dirige a la Tierra. La transmisión contiene datos vitales sobre las dimensiones, la velocidad y la potencia de fuego de cada nave espacial alienígena . Tu trabajo consiste en analizar estos datos para proporcionar a la EDO la información necesaria para preparar las defensas de la Tierra.

Ejercicio:

La transmisión es una matriz (un arreglo) que contiene datos para cada nave espacial. Cada fila representa una nave espacial, con valores en el siguiente orden: longitud (en metros), anchura (en metros), altura (en metros), velocidad (en km/s) y potencia de las armas (en una escala del 1 al 10).

```python
import numpy as np


fleet_data = np.array([
    [120, 40, 25, 2.5, 10],
    [85, 60, 30, 5.0, 5],
    [100, 50, 35, 4.5, 7],
    [120, 40, 25, 2.5, 9],
    [150, 50, 30, 3.0, 10],
], dtype=np.float32)
fleet_data
```

1. Calcula e imprime el número total de naves espaciales.

```python
num_naves = fleet_data.shape[0]
print(f"El número de naves es: {num_naves}")

nu_naves = len(fleet_data)
print(f"El número de naves es: {nu_naves}")
```

2. Calcula e imprime el tamaño (volumen) promedio de la nave espacial.

```python
vol_total_naves = np.prod(fleet_data[:, :3], axis=1)
print(f"El volumen total de cada nave es: {vol_total_naves} metros cúbicos")

vol_medio_naves = np.mean(vol_total_naves)
print(f"El volumen promedio de las naves es: {vol_medio_naves} metros cúbicos")
```

3. Identifica e imprime la velocidad de la nave espacial más rápida.

```python
velocidades_n = fleet_data[:, 3]
print(f"Las velocidades de las naves son: {velocidades_n} km/s")

velocidad_max = np.max(velocidades_n)
print(f"La velocidad máxima de las naves es: {velocidad_max} km/s")
```

4. Determina e imprime el número de naves espaciales con la máxima potencia de armamento.

```python
potencia_armas_n = fleet_data[:, 4]
print(f"Las potencias de las armas de las naves son: {potencia_armas_n}")

max_potencia_armas = np.max(potencia_armas_n)
print(f"La potencia máxima de las armas de las naves es: {max_potencia_armas}")

#con sum
numero_naves_max_potencia = np.sum(potencia_armas_n == max_potencia_armas)
print(f"El número de naves con la potencia máxima de armas es: {numero_naves_max_potencia} (con sum)")

#con len
naves_max_p = potencia_armas_n == max_potencia_armas
print(f"Las naves con la potencia máxima de armas son: {naves_max_p}")
num_naves_max_p = len(fleet_data[naves_max_p])
print(f"El número de naves con la potencia máxima de armas es: {num_naves_max_p} (con len)")
```

5. Calcula e imprime la velocidad media de las naves espaciales con la máxima potencia de armamento.

```python
naves_mpa = fleet_data[naves_max_p]
print(f"Las naves con la potencia máxima de armas son: {naves_mpa}")

vel_med_naves_mpa = np.mean(naves_mpa[:, 3])
print(f"La velocidad promedio de las naves con la potencia máxima de armas es: {vel_med_naves_mpa} km/s")
```
